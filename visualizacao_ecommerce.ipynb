```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('ecommerce_estatistica.csv')

df.head()
```

```python
df.info()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Preço'], bins=20)
plt.title('Distribuição dos Preços')
plt.xlabel('Preço')
plt.ylabel('Frequência')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Nota'], df['Qtd_Vendidos_Cod'])
plt.title('Nota x Quantidade Vendida')
plt.xlabel('Nota')
plt.ylabel('Quantidade Vendida')
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='viridis')
plt.title('Mapa de Calor das Correlações')
plt.show()
```

```python
plt.figure(figsize=(10,5))
df['Marca'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Marcas com Mais Produtos')
plt.xlabel('Marca')
plt.ylabel('Quantidade')
plt.show()
```

```python
plt.figure(figsize=(7,7))
df['Temporada'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Distribuição por Temporada')
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['Preço'], fill=True)
plt.title('Densidade dos Preços')
plt.xlabel('Preço')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.regplot(x='Nota', y='Qtd_Vendidos_Cod', data=df)
plt.title('Regressão: Nota x Quantidade Vendida')
plt.xlabel('Nota')
plt.ylabel('Quantidade Vendida')
plt.show()
```
